```python
import os

PROJECT_DIR = "/home/pi/SortiFy_Dashboard"   # <- change if your folder is different
os.chdir(PROJECT_DIR)

print("Current working directory:", os.getcwd())
```

```python
import subprocess, time

# If we already have a running backend_proc from a previous run, try to stop it first
try:
    backend_proc.terminate()
except NameError:
    pass
except Exception:
    pass

backend_proc = subprocess.Popen(
    ["python", "integrated_backend.py"],
    cwd=PROJECT_DIR,
)

time.sleep(5)  # give it a few seconds to start
print("Backend started on http://0.0.0.0:8000  (PID:", backend_proc.pid, ")")
```

```python
# Stop old frontend, if any
try:
    frontend_proc.terminate()
except NameError:
    pass
except Exception:
    pass

frontend_proc = subprocess.Popen(
    ["python", "-m", "http.server", "5500"],
    cwd=PROJECT_DIR,
)

print("Frontend started on http://<Pi-IP>:5500/index.html  (PID:", frontend_proc.pid, ")")
```

```python
for name, proc in [("backend", "backend_proc"), ("frontend", "frontend_proc")]:
    try:
        p = globals()[proc]
        p.terminate()
        print(f"Stopped {name} (PID {p.pid})")
    except KeyError:
        print(f"{name.capitalize()} was not started in this session.")
    except Exception as e:
        print(f"Error stopping {name}:", e)
```
